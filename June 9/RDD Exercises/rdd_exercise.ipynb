```python
total=rdd.count()
unique=rdd.distinct().count()
freq=rdd.countByValue()
most=max(freq.items(),key=lambda x:x[1])
least=min(freq.items(),key=lambda x:x[1])
login_percent=(rdd.filter(lambda x:x=="LOGIN").count()/total)*100
video_percent=(rdd.filter(lambda x:x=="VIDEO_PLAY").count()/total)*100

print("Total Activities:",total)
print("Unique Activities:",unique)
print("Most Common Activity:",most)
print("Least Common Activity:",least)
print("LOGIN Percentage:",login_percent)
print("VIDEO_PLAY Percentage:",video_percent)
print("Activity Frequencies:",freq)
```
