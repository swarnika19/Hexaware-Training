```python
total_transactions=rdd.count()
unique_transactions=rdd.distinct().count()
transaction_freq=rdd.countByValue()
highest_transaction=max(transaction_freq.items(),key=lambda x:x[1])
lowest_transaction=min(transaction_freq.items(),key=lambda x:x[1])
deposit_percentage=(rdd.filter(lambda x:x=="DEPOSIT").count()/total_transactions)*100
transfer_percentage=(rdd.filter(lambda x:x=="TRANSFER").count()/total_transactions)*100
withdraw_percentage=(rdd.filter(lambda x:x=="WITHDRAW").count()/total_transactions)*100
total_money=amount_rdd.reduce(lambda x,y:x+y)
average_money=total_money/amount_rdd.count()
high_value=amount_rdd.filter(lambda x:x>20000).collect()
low_value=amount_rdd.filter(lambda x:x<1000).collect()

print("BANK TRANSACTION DASHBOARD")
print("Total Banking Activities:",total_transactions)
print("Unique Transaction Types:",unique_transactions)
print("Transaction Frequencies:",transaction_freq)
print("Highest Frequency Transaction:",highest_transaction)
print("Lowest Frequency Transaction:",lowest_transaction)
print("Deposit Percentage:",deposit_percentage)
print("Transfer Percentage:",transfer_percentage)
print("Withdraw Percentage:",withdraw_percentage)
print("Total Money Processed:",total_money)
print("Average Money Processed:",average_money)
print("High Value Transactions:",high_value)
print("Low Value Transactions:",low_value)
```
