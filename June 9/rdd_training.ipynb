```python
# create sparksession and RDD app
from pyspark.sql import SparkSession
spark = SparkSession.builder \
.appName("RDD") \
.getOrCreate()
```

```python
# dataset is parallelized
numbers = [10,20,30,40,50]
rdd = spark.sparkContext.parallelize(numbers)
print(rdd.collect())
```

```python
# count the total number of items in the rdd dataset
rdd.count()
```

```python
# print the first three items in rdd dataset
rdd.take(3)
```

```python
# multply each number by 10
numbers = [10,20,30,40,50]
rdd = spark.sparkContext.parallelize(numbers)

# only transformation happens
result = rdd.map(
    lambda x: x* 10
)

# action to rdd (lazy evaluation)
print(result.collect())
```

```python
# square each number
numbers = [1,2,3,4,5]
rdd = spark.sparkContext.parallelize(numbers)
result = rdd.map(
    lambda x: x* x
)
print(result.collect())
```

```python
# even numbers
numbers = [1,2,3,4,5,6,7,8]
rdd = spark.sparkContext.parallelize(numbers)
result = rdd.filter(
    lambda x: x%2==0
)

print(result.collect())
```

```python
# find unique items
numbers = [10,20,20,30,40,40,50]
rdd = spark.sparkContext.parallelize(numbers)
result = rdd.distinct()
print(result.collect())
```

```python
# union of two rdd datasets
rdd1 = spark.sparkContext.parallelize(
    [10,20,30]
)
rdd2 = spark.sparkContext.parallelize(
    [40,50,60]
)
result = rdd1.union(rdd2)
print(result.collect())
```

```python
numbers = [40,10,60,20,50]

rdd = spark.sparkContext.parallelize(numbers)

result = rdd.sortBy(
    lambda x: x
)

print(result.collect())
```

```python
numbers = [10,20,30,40]

rdd = spark.sparkContext.parallelize(numbers)

result = rdd.reduce(
    lambda x,y: x+y
)

print(result)
```
