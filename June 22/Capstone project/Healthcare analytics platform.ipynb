```python
# DATA CLEANING AND TRANSFORMATION

from pyspark.sql.functions import col, when, count, to_date, month, year
patients_df.select([count(when(col(c).isNull(), c)).alias(c) for c in patients_df.columns]).show()
doctors_df.select([count(when(col(c).isNull(), c)).alias(c) for c in doctors_df.columns]).show()
appointments_df.select([count(when(col(c).isNull(), c)).alias(c) for c in appointments_df.columns]).show()
preferences_df.select([count(when(col(c).isNull(), c)).alias(c) for c in preferences_df.columns]).show()
patients_df = patients_df.fillna({"insurance_status": "Unknown", "gender": "Unknown"})
doctors_df = doctors_df.fillna({"consultation_fee": 0})
appointments_df = appointments_df.fillna({"bill_amount": 0, "status": "Unknown", "diagnosis": "Not Specified"})
preferences_df = preferences_df.fillna({"preferred_hospital": "Not Specified", "phone": "N/A", "email": "N/A"})
print("Missing values handled!")
```

```python
silver_df = silver_df.withColumn("appointment_date", to_date(col("appointment_date"), "yyyy-MM-dd"))
silver_df = silver_df.withColumn("appointment_month", month(col("appointment_date")))
silver_df = silver_df.withColumn("appointment_year", year(col("appointment_date")))
display(silver_df.select("appointment_id", "appointment_date", "appointment_month", "appointment_year"))
```

```python
silver_df = silver_df.withColumn("age_group", when(col("age") >= 50, "Senior").when(col("age") >= 30, "Adult").otherwise("Young"))
display(silver_df.select("patient_name", "age", "age_group"))
```

```python
# WINDOW FUNCTIONS

from pyspark.sql.functions import sum, rank, dense_rank
from pyspark.sql.window import Window
doctor_window = Window.orderBy(sum("final_bill").desc())
doctor_revenue_df = silver_df.groupBy("doctor_id", "doctor_name").agg(sum("final_bill").alias("total_revenue"))
doctor_revenue_df = doctor_revenue_df.withColumn("doctor_rank", rank().over(Window.orderBy(col("total_revenue").desc())))
display(doctor_revenue_df)
```

```python
dept_revenue_df = silver_df.groupBy("department").agg(sum("final_bill").alias("total_revenue"))
dept_revenue_df = dept_revenue_df.withColumn("dept_rank", rank().over(Window.orderBy(col("total_revenue").desc())))
display(dept_revenue_df)
```

```python
patient_billing_df = silver_df.groupBy("patient_id", "patient_name").agg(sum("final_bill").alias("total_billed"))
patient_billing_df = patient_billing_df.withColumn("billing_rank", rank().over(Window.orderBy(col("total_billed").desc())))
top3_patients_df = patient_billing_df.filter(col("billing_rank") <= 3)
display(top3_patients_df)
```

```python
dept_doctor_window = Window.partitionBy("department").orderBy(col("total_revenue").desc())
doctor_dept_df = silver_df.groupBy("department", "doctor_name").agg(sum("final_bill").alias("total_revenue"))
doctor_dept_df = doctor_dept_df.withColumn("rank_in_dept", rank().over(dept_doctor_window))
top_doctor_per_dept = doctor_dept_df.filter(col("rank_in_dept") == 1)
display(top_doctor_per_dept)
```

```python
from pyspark.sql.functions import sum as spark_sum
running_window = Window.orderBy("appointment_date").rowsBetween(Window.unboundedPreceding, Window.currentRow)
running_df = silver_df.groupBy("appointment_date").agg(spark_sum("final_bill").alias("daily_revenue"))
running_df = running_df.withColumn("running_revenue", spark_sum("daily_revenue").over(running_window))
display(running_df.orderBy("appointment_date"))
```

```python
from delta.tables import DeltaTable
updated_patients = [
    ("P101", "Rahul Sharma", "Pune", "Maharashtra", 35, "Male", "Inactive"),
    ("P103", "Amit Kumar", "Delhi", "Delhi", 42, "Male", "Active"),
    ("P109", "Karthik Raj", "Chennai", "Tamil Nadu", 30, "Male", "Active")
]
updated_schema = ["patient_id","patient_name","city","state","age","gender","insurance_status"]
updated_df = spark.createDataFrame(updated_patients, updated_schema)
delta_table = DeltaTable.forName(spark, "my_databricks_azure_day6.healthcare_gold.gold_patients_sql")
delta_table.alias("target").merge(
    updated_df.alias("source"),
    "target.patient_id = source.patient_id"
).whenMatchedUpdate(set={
    "city": "source.city",
    "state": "source.state",
    "insurance_status": "source.insurance_status"
}).whenNotMatchedInsertAll().execute()
spark.sql("SELECT * FROM my_databricks_azure_day6.healthcare_gold.gold_patients_sql").show()
```
