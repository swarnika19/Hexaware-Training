```python
# CLEANING AND TRANSFORMATION

from pyspark.sql.functions import col, when, count, to_date, month, year
customers_df.select([count(when(col(c).isNull(), c)).alias(c) for c in customers_df.columns]).show()
products_df.select([count(when(col(c).isNull(), c)).alias(c) for c in products_df.columns]).show()
orders_df.select([count(when(col(c).isNull(), c)).alias(c) for c in orders_df.columns]).show()
preferences_df.select([count(when(col(c).isNull(), c)).alias(c) for c in preferences_df.columns]).show()
customers_df = customers_df.fillna({"city": "Unknown", "state": "Unknown", "customer_type": "Regular"})
products_df = products_df.fillna({"category": "Unknown", "unit_price": 0})
orders_df = orders_df.fillna({"status": "Unknown", "quantity": 0})
preferences_df = preferences_df.fillna({"preferred_channel": "Unknown", "loyalty_tier": "Unknown", "loyalty_points": 0})
print("Null values handled!")
```

```python
# WINDOW FUNCTIONS

from pyspark.sql.functions import sum, rank
from pyspark.sql.window import Window
product_revenue_df = silver_df.groupBy("product_id", "product_name").agg(sum("revenue").alias("total_revenue"))
product_revenue_df = product_revenue_df.withColumn("product_rank", rank().over(Window.orderBy(col("total_revenue").desc())))
display(product_revenue_df)
```

```python
customer_revenue_df = silver_df.groupBy("customer_id", "customer_name").agg(sum("revenue").alias("total_revenue"))
customer_revenue_df = customer_revenue_df.withColumn("customer_rank", rank().over(Window.orderBy(col("total_revenue").desc())))
display(customer_revenue_df)
```

```python
top3_customers_df = customer_revenue_df.filter(col("customer_rank") <= 3)
display(top3_customers_df)
```

```python
category_window = Window.partitionBy("category").orderBy(col("total_revenue").desc())
product_category_df = silver_df.groupBy("category", "product_name").agg(sum("revenue").alias("total_revenue"))
product_category_df = product_category_df.withColumn("rank_in_category", rank().over(category_window))
top_product_per_category = product_category_df.filter(col("rank_in_category") == 1)
display(top_product_per_category)
```

```python
from pyspark.sql.functions import sum as spark_sum
running_window = Window.orderBy("order_date").rowsBetween(Window.unboundedPreceding, Window.currentRow)
running_df = silver_df.groupBy("order_date").agg(spark_sum("revenue").alias("daily_revenue"))
running_df = running_df.withColumn("running_revenue", spark_sum("daily_revenue").over(running_window))
display(running_df.orderBy("order_date"))
```

```python
from delta.tables import DeltaTable
updated_customers = [
    ("C101", "Rahul Sharma", "Mumbai", "Maharashtra", "Regular"),
    ("C103", "Amit Kumar", "Pune", "Maharashtra", "Premium"),
    ("C109", "Kiran Raj", "Chennai", "Tamil Nadu", "Regular")
]
updated_schema = ["customer_id", "customer_name", "city", "state", "customer_type"]
updated_df = spark.createDataFrame(updated_customers, updated_schema)
delta_table = DeltaTable.forName(spark, "`my-databricks-azure-day6`.retail_gold.gold_customers_sql")
delta_table.alias("target").merge(
    updated_df.alias("source"),
    "target.customer_id = source.customer_id"
).whenMatchedUpdate(set={
    "city": "source.city",
    "state": "source.state",
    "customer_type": "source.customer_type"
}).whenNotMatchedInsertAll().execute()
spark.sql("SELECT * FROM `my-databricks-azure-day6`.retail_gold.gold_customers_sql").show()
```

```python
# VISUALIZATION

import matplotlib.pyplot as plt
city_df = silver_df.groupBy("city").agg(sum("revenue").alias("total_revenue")).orderBy(col("total_revenue").desc()).toPandas()
plt.figure(figsize=(10,6))
plt.bar(city_df["city"], city_df["total_revenue"], color="steelblue")
plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_df = silver_df.groupBy("category").agg(sum("revenue").alias("total_revenue")).orderBy(col("total_revenue").desc()).toPandas()
plt.figure(figsize=(10,6))
plt.bar(category_df["category"], category_df["total_revenue"], color="coral")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
status_df = silver_df.groupBy("status").count().toPandas()
plt.figure(figsize=(8,8))
plt.pie(status_df["count"], labels=status_df["status"], autopct="%1.1f%%", colors=["green","orange","red"])
plt.title("Orders by Status")
plt.tight_layout()
plt.show()
```

```python
product_df = silver_df.groupBy("product_name").agg(sum("revenue").alias("total_revenue")).orderBy(col("total_revenue").desc()).toPandas()
plt.figure(figsize=(10,6))
plt.barh(product_df["product_name"], product_df["total_revenue"], color="mediumpurple")
plt.title("Top Products by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
date_df = silver_df.groupBy("order_date").agg(sum("revenue").alias("total_revenue")).orderBy("order_date").toPandas()
plt.figure(figsize=(10,6))
plt.plot(date_df["order_date"].astype(str), date_df["total_revenue"], marker="o", color="teal", linewidth=2)
plt.title("Revenue Trend by Date")
plt.xlabel("Order Date")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
