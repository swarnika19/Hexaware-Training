```python
w_city = Window.partitionBy("city").orderBy(col("price").desc())
df.withColumn("rank", rank().over(w_city)).filter(col("rank") == 1).show()
```

```python
w_cat = Window.partitionBy("category").orderBy(col("price").desc())
df.withColumn("rank", rank().over(w_cat)).filter(col("rank") == 1).show()
```

```python
w_running = Window.orderBy(col("price").desc()).rowsBetween(Window.unboundedPreceding, Window.currentRow)
df.withColumn("running_total", sum(col("price")).over(w_running)).show()
```

```python
# Mini ETL Exercises
source_df = spark.createDataFrame(sales_data, columns)
source_df.show()
```

```python
etl_df = source_df.withColumn("total_amount", col("quantity") * col("price")).withColumn("tax", col("total_amount") * 0.05).withColumn("final_amount", col("total_amount") + col("tax"))
etl_df.show()
```

```python
completed_df = etl_df.filter(col("status") == "Completed")
completed_df.show()
```

```python
revenue_df = completed_df.withColumn("revenue", col("quantity") * col("price"))
revenue_df.show()
```

```python
city_report = revenue_df.groupBy("city").sum("revenue").withColumnRenamed("sum(revenue)", "total_revenue").orderBy(col("total_revenue").desc())
city_report.show()
```

```python
category_report = revenue_df.groupBy("category").sum("revenue").withColumnRenamed("sum(revenue)", "total_revenue").orderBy(col("total_revenue").desc())
category_report.show()
```

```python
revenue_df.write.mode("overwrite").saveAsTable("sales_output")
```

```python
parquet_df = spark.table("sales_output")
parquet_df.show()
```

```python
parquet_df.createOrReplaceTempView("sales_parquet")
spark.sql("SELECT * FROM sales_parquet").show()
```
