```python
# Window Function Exercises

from pyspark.sql.functions import rank, dense_rank, row_number, sum, lag, lead
from pyspark.sql.window import Window

doctor_revenue = doctors_df.join(visits_df, "doctor_id", "inner").groupBy("doctor_id", "doctor_name", "specialization", "city").sum("bill_amount").withColumnRenamed("sum(bill_amount)", "revenue")

w = Window.orderBy(col("revenue").desc())
doctor_revenue.withColumn("rank", rank().over(w)).show()
```

```python
doctor_revenue.withColumn("dense_rank", dense_rank().over(w)).show()
doctor_revenue.withColumn("row_number", row_number().over(w)).show()
```

```python
doctor_revenue.withColumn("rank", rank().over(w)).filter(col("rank") == 1).show()
doctor_revenue.withColumn("rank", rank().over(w)).filter(col("rank") <= 3).show()
w_spec = Window.partitionBy("specialization").orderBy(col("revenue").desc())
doctor_revenue.withColumn("rank", rank().over(w_spec)).filter(col("rank") == 1).show()
```

```python
doctor_revenue.withColumn("rank", rank().over(w_spec)).filter(col("rank") <= 2).show()
```

```python
w_running = Window.orderBy(col("revenue").desc()).rowsBetween(Window.unboundedPreceding, Window.currentRow)
doctor_revenue.withColumn("running_revenue", sum(col("revenue")).over(w_running)).show()
```

```python
doctor_revenue.withColumn("prev_revenue", lag(col("revenue"), 1).over(w)).show()
```

```python
doctor_revenue.withColumn("next_revenue", lead(col("revenue"), 1).over(w)).show()
```

```python
doctor_revenue.withColumn("prev_revenue", lag(col("revenue"), 1).over(w)).withColumn("diff_with_prev", col("revenue") - col("prev_revenue")).show()
```

```python
doctor_revenue.withColumn("next_revenue", lead(col("revenue"), 1).over(w)).withColumn("diff_with_next", col("revenue") - col("next_revenue")).show()
```

```python
w_city = Window.partitionBy("city").orderBy(col("revenue").desc())
doctor_revenue.withColumn("rank", rank().over(w_city)).filter(col("rank") == 1).show()
```

```python
w_city_asc = Window.partitionBy("city").orderBy(col("revenue").asc())
doctor_revenue.withColumn("rank", rank().over(w_city_asc)).filter(col("rank") == 1).show()
```

```python
w_running = Window.orderBy(col("revenue").desc()).rowsBetween(Window.unboundedPreceding, Window.currentRow)
doctor_revenue.withColumn("rank", rank().over(w)).withColumn("dense_rank", dense_rank().over(w)).withColumn("row_number", row_number().over(w)).withColumn("running_revenue", sum(col("revenue")).over(w_running)).orderBy(col("revenue").desc()).show()
```

```python
# ETL exercises

doctors_df = spark.read.option("header", True).option("inferSchema", True).csv("file:/Workspace/Users/azuser7227_mml.local@karthikirisoutlook.onmicrosoft.com/doctors.csv")
visits_df = spark.read.option("header", True).option("inferSchema", True).csv("file:/Workspace/Users/azuser7227_mml.local@karthikirisoutlook.onmicrosoft.com/visits.csv")
hospital_df = spark.read.option("multiline", True).json("file:/Workspace/Users/azuser7227_mml.local@karthikirisoutlook.onmicrosoft.com/hospital_config.json")
doctors_df.show()
visits_df.show()
hospital_df.show()
```

```python
visits_clean = visits_df.filter(col("doctor_id") != "D120").fillna({"bill_amount": 0})
visits_clean.show()
```

```python
from pyspark.sql.functions import explode

hospital_flat = hospital_df.withColumn("phone", col("contact.phone")).withColumn("email", col("contact.email")).withColumn("service", explode(col("services"))).drop("contact", "services").fillna({"phone": "Not Available", "email": "Not Available"})
hospital_flat.show()
```

```python
joined_df = doctors_df.join(visits_clean, "doctor_id", "inner")
joined_df.show()
```

```python
revenue_df = joined_df.withColumn("tax", col("bill_amount") * 0.05).withColumn("final_bill", col("bill_amount") + col("tax")).withColumn("discount", col("bill_amount") * 0.10).withColumn("net_amount", col("bill_amount") - col("discount"))
revenue_df.show()
```

```python
from pyspark.sql.functions import rank, sum
from pyspark.sql.window import Window

doctor_revenue = revenue_df.groupBy("doctor_id", "doctor_name", "specialization", "city").sum("bill_amount").withColumnRenamed("sum(bill_amount)", "total_revenue")

w = Window.orderBy(col("total_revenue").desc())
doctor_ranked = doctor_revenue.withColumn("rank", rank().over(w))
doctor_ranked.show()
```

```python
from pyspark.sql.functions import count, avg, max, min
spec_summary = revenue_df.groupBy("specialization").agg(count("visit_id").alias("total_visits"), sum("bill_amount").alias("total_revenue"), avg("bill_amount").alias("avg_revenue"), max("bill_amount").alias("max_bill"), min("bill_amount").alias("min_bill"))
spec_summary.show()
```

```python
revenue_df.write.mode("overwrite").saveAsTable("silver_doctor_visits")
doctor_ranked.write.mode("overwrite").saveAsTable("silver_doctor_ranking")
spec_summary.write.mode("overwrite").saveAsTable("silver_specialization_summary")
```

```python
city_report = revenue_df.groupBy("city").agg(count("visit_id").alias("total_visits"), sum("bill_amount").alias("total_revenue"), avg("bill_amount").alias("avg_revenue")).orderBy(col("total_revenue").desc())
city_report.show()

top_doctors = doctor_ranked.filter(col("rank") <= 3)
top_doctors.show()

payment_report = visits_clean.groupBy("payment_status").agg(count("visit_id").alias("total_visits"), sum("bill_amount").alias("total_amount"))
payment_report.show()
```

```python
dashboard_df = revenue_df.groupBy("doctor_id", "doctor_name", "specialization", "city").agg(count("visit_id").alias("total_visits"), sum("bill_amount").alias("total_revenue"), avg("bill_amount").alias("avg_revenue"), max("bill_amount").alias("max_bill"), min("bill_amount").alias("min_bill"), sum("tax").alias("total_tax"), sum("final_bill").alias("total_final_bill"), sum("discount").alias("total_discount"), sum("net_amount").alias("total_net_amount")).withColumn("rank", rank().over(Window.orderBy(col("total_revenue").desc()))).orderBy(col("total_revenue").desc())

dashboard_df.show()
dashboard_df.write.mode("overwrite").saveAsTable("gold_hospital_dashboard")
```
