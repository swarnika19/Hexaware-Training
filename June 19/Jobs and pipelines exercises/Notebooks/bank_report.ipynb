```python
# Task 2: Create temp view
bank_df.createOrReplaceTempView("bank_view")

# Calculate total balance by city using SQL
city_report = spark.sql("""
    SELECT
        city,
        COUNT(account_id)           AS total_accounts,
        SUM(balance)                AS total_balance,
        ROUND(AVG(balance), 2)      AS avg_balance,
        MAX(balance)                AS max_balance,
        MIN(balance)                AS min_balance
    FROM bank_view
    GROUP BY city
    ORDER BY total_balance DESC
""")

print("=== Total Balance by City ===")
display(city_report)
```

```python
# Task 3: Display account type breakdown
type_report = spark.sql("""
    SELECT
        account_type,
        COUNT(account_id)           AS total_accounts,
        SUM(balance)                AS total_balance,
        ROUND(AVG(balance), 2)      AS avg_balance
    FROM bank_view
    GROUP BY account_type
    ORDER BY total_balance DESC
""")

print("=== Balance by Account Type ===")
display(type_report)

print("Daily Banking Report Completed Successfully!")
```
