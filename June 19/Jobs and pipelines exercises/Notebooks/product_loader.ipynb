```python
from pyspark.sql import Row
# Task 1: Create product data
product_data = [
    Row(product_id=1, product_name="Laptop",     category="Electronics", price=75000.00),
    Row(product_id=2, product_name="Mouse",       category="Electronics", price=1500.00),
    Row(product_id=3, product_name="Desk Chair",  category="Furniture",   price=12000.00),
    Row(product_id=4, product_name="Notebook",    category="Stationery",  price=200.00),
    Row(product_id=5, product_name="Monitor",     category="Electronics", price=25000.00),
    Row(product_id=6, product_name="Keyboard",    category="Electronics", price=3000.00),
    Row(product_id=7, product_name="Bookshelf",   category="Furniture",   price=8000.00),
    Row(product_id=8, product_name="Pen Set",     category="Stationery",  price=150.00),
]
# Task 2: Create a DataFrame
df = spark.createDataFrame(product_data)
print("=== All Products ===")
display(df)
```

```python
# Create a Temp View
df.createOrReplaceTempView("products_view")
print("Temp view 'products_view' created successfully.")
```

```python
#Calculate average product price using SQL (by category)
avg_price_df = spark.sql("""
    SELECT 
        category,
        COUNT(product_id)    AS total_products,
        ROUND(AVG(price), 2) AS avg_price,
        MIN(price)           AS min_price,
        MAX(price)           AS max_price
    FROM products_view
    GROUP BY category
    ORDER BY avg_price DESC
""")

#Display the results
print("=== Average Price by Category ===")
display(avg_price_df)
```
