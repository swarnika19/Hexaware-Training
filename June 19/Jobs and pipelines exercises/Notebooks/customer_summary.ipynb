```python
# Read customer_table
customer_df = spark.read.format("delta").table("customer_table")
print("=== customer_table Loaded ===")
display(customer_df)
```

```python
# Create temp view
customer_df.createOrReplaceTempView("customer_view")

# Calculate customers by city using SQL
city_summary = spark.sql("""
    SELECT 
        city,
        COUNT(customer_id)        AS total_customers,
        ROUND(AVG(spend), 2)      AS avg_spend,
        SUM(spend)                AS total_spend
    FROM customer_view
    GROUP BY city
    ORDER BY total_customers DESC
""")

print("=== Customer Summary by City ===")
display(city_summary)
```
