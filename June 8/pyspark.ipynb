```python
# Create a new spark session
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySpark Day 1") \
    .getOrCreate()
```

```python
spark
```

```python
# Raw data
data = [
    (101, "Rahul Sharma", "Data Engineering", 75000, "Hyderabad"),
    (102, "Priya Reddy", "AI Engineering", 85000, "Bangalore"),
    (103, "Amit Kumar", "Data Engineering", 65000, "Mumbai"),
    (104, "Sneha Patel", "Data Science", 95000, "Chennai"),
    (105, "Farhan Ali", "Cloud Engineering", 80000, "Delhi")
]
```

```python
# Naming the columns of df
columns = [
    "employee_id",
    "name",
    "department",
    "salary",
    "city"
]
```

```python
# Convert raw data to df and display
df = spark.createDataFrame(data, columns)
df.show()
```

```python
# Print the schema of dataframe
df.printSchema()
```

```python
# Select based on cols
df.select("name", "salary").show()
```

```python
# filter salary more than 80000
df.filter(df.salary > 80000).show()
```

```python
# print rows where city is hyd
df.filter(df.city == "Hyderabad").show()
```

```python
# sort based on salary in ascending order
df.orderBy("salary").show()
```

```python
# sort salary in descending order
df.orderBy(df.salary.desc()).show()
```

```python
# add new column in df using withColumn to increase salary by 10 percentage
from pyspark.sql.functions import col
df_bonus = df.withColumn(
    "bonus",
    col("salary") * 0.10
)
df_bonus.show()
```

```python
# rename salary col to monthly_salary
df_renamed = df.withColumnRenamed(
    "salary",
    "monthly_salary"
)
df_renamed.show()
```

```python
# group by rows based on department with their count
df.groupBy("department").count().show()
```

```python
df2 = df.withColumn(
    "salary_category",
    F.when(F.col("salary") >= 85000, "High")
     .when(F.col("salary") >= 75000, "Medium")
     .otherwise("Low")
)

df2.groupBy("salary_category").count().show()
```
