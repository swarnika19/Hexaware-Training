```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import (
    col,
    when,
    count,
    sum,
    avg
)
```

```python
quality_df = bookings_df.withColumn(
    "data_quality_status",
    when(
        col("customer_name").isNull() |
        col("city").isNull() |
        col("service_type").isNull() |
        col("provider").isNull() |
        col("booking_amount").isNull() |
        col("booking_status").isNull() |
        col("payment_mode").isNull(),
        "Incomplete"
    ).otherwise("Complete")
)

quality_df.show()
```

```python
quality_df.groupBy("data_quality_status").count().show()
```

```python
quality_df.filter(
    col("data_quality_status") == "Complete"
).show()
```

```python
quality_df.filter(
    col("data_quality_status") == "Incomplete"
).show()
```

```python
tax_df = bookings_df.withColumn(
    "tax",
    col("booking_amount") * 0.05
)

tax_df.show()
```

```python
final_df = tax_df.withColumn(
    "final_amount",
    col("booking_amount") + col("tax")
)

final_df.show()
```

```python
bookings_df.filter(
    col("booking_status") == "Confirmed"
).agg(
    sum("booking_amount").alias("confirmed_revenue")
).show()
```

```python
avg_with_zero = bookings_df.na.fill(
    {"booking_amount": 0}
).agg(
    avg("booking_amount").alias("avg_with_zero")
)

avg_without_null = bookings_df.na.drop(
    subset=["booking_amount"]
).agg(
    avg("booking_amount").alias("avg_without_null")
)

avg_with_zero.show()
avg_without_null.show()
```
