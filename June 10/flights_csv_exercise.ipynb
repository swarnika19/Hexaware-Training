```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
.appName("Working with files") \
.getOrCreate()
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
flights_df = spark.read.csv(
    "flights.csv",
    header = True,
    inferSchema = True
)
flights_df.show()
flights_df.printSchema()
```

```python
flights_df.select("passenger_name", "from_city", "to_city").show()
```

```python
from pyspark.sql.functions import col, sum, avg, max, min, when
```

```python
flights_df.filter(col("status") == "Confirmed").show()
```

```python
flights_df.filter(col("status") == "Cancelled").show()
```

```python
flights_df.filter(col("status") == "Pending").show()
```

```python
flights_df.filter(col("ticket_price") > 20000).show()
```

```python
flights_df.filter(col("to_city").isin("Dubai", "Singapore", "London")).show()
```

```python
flights_df.count()
```

```python
flights_df.groupBy("airline").count().show()
```

```python
flights_df.groupBy("status").count().show()
```

```python
flights_df.filter(col("status") == "Confirmed").agg(sum("ticket_price").alias("total_revenue")).show()
```

```python
flights_df.groupBy("airline").agg(avg("ticket_price").alias("average_ticket_price")).show()
```

```python
flights_df.agg(max("ticket_price").alias("hightest_ticket_price")).show()
```

```python
flights_df.agg(min("ticket_price").alias("lowest_ticket_price")).show()
```

```python
flights_df.orderBy(col("ticket_price").desc()).show()
```

```python
flights_df = flights_df.withColumn("tax", col("ticket_price") * 0.05)
flights_df.show()
```

```python
flights_df = flights_df.withColumn("final_price", col("ticket_price") + col("tax"))
flights_df.show()
```

```python
flights_df = flights_df.withColumn(
    "price_category",
    when(col("ticket_price") >= 30000, "Premium")
    .when(col("ticket_price") >= 10000, "Standard")
    .otherwise("Budget")
)
flights_df.show()
```

```python
flights_df.groupBy("price_category").count().show()
```

```python
confirmed_df = flights_df.filter(col("status") == "Confirmed")
confirmed_df.write.mode("overwrite").parquet("confirmed_flights.parquet")
```
